## Feature Engineering

### 1. BMI Winsorization

```python
bmi_low, bmi_high = df_clean["BMI"].quantile([0.01, 0.99])
df_clean["BMI_capped"] = df_clean["BMI"].clip(bmi_low, bmi_high)
```

### 2. General Health Mapping

```python
genhlth_map = {1: "Excellent", 2: "Very Good", 3: "Good", 4: "Fair", 5: "Poor"}
df_clean["GenHlth_Factor"] = df_clean["GenHlth"].map(genhlth_map)
```

### 3. Filter Invalid Health Values

```python
df_clean = df_clean[(df_clean["MentHlth"] <= 30) & (df_clean["PhysHlth"] <= 30)]
```

### 4. BMI Categories

```python
df_clean["BMI_Category"] = pd.cut(
    df_clean["BMI"],
    bins=[0, 18.5, 25, 30, float("inf")],
    labels=["Underweight", "Normal", "Overweight", "Obese"]
)
```

### 5. Age Group Mapping

```python
age_labels = [
    "18-24","25-29","30-34","35-39","40-44","45-49",
    "50-54","55-59","60-64","65-69","70-74","75-79","80+"
]
df_clean["AgeGroup"] = df_clean["Age"].astype(int).apply(lambda x: age_labels[x-1])
```

### 6. Age Bands

```python
df_clean["AgeBand"] = pd.cut(
    df_clean["Age"],
    bins=[0,3,6,9,11,14],
    labels=["18-34","35-49","50-64","65-74","75+"]
)
```

### 7. Risk Score

```python
df_clean["RiskScore"] = (
    df_clean["Smoker"] +
    df_clean["HvyAlcoholConsump"] +
    (1 - df_clean["PhysActivity"]) +
    (1 - df_clean["Fruits"]) +
    (1 - df_clean["Veggies"])
)
```

### 8. Disease Count

```python
df_clean["DiseaseCount"] = (
    df_clean["HighBP"] + df_clean["HighChol"] +
    df_clean["Diabetes"] + df_clean["Stroke"]
)
```

### 9. Health Stress Index

```python
df_clean["HealthStressIndex"] = df_clean["MentHlth"] + df_clean["PhysHlth"]
```

### 10. Healthcare Access Score

```python
df_clean["HealthcareScore"] = df_clean["AnyHealthcare"] + (1 - df_clean["NoDocbcCost"])
```

### 11. Obesity Flag

```python
df_clean["ObeseFlag"] = (df_clean["BMI"] >= 30).astype(int)
```

### 12. Lifestyle Profile

```python
df_clean["LifestyleProfile"] = pd.cut(
    df_clean["RiskScore"],
    bins=[-1,1,3,10],
    labels=["Healthy","ModerateRisk","HighRisk"]
)
```
